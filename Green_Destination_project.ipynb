```python

# Encoding Attrition
df['Attrition_Encoded'] = df['Attrition'].map({'Yes':1,'No':0})

# Converting Categorical text columns to category type
for col in df.select_dtypes('object').columns:
    df[col] = df[col].astype('category')

df['IncomeBin'] = pd.cut(df['MonthlyIncome'],
                            bins=[0, 3000, 6000, 9000, 12000, 15000, 20000],
                            labels=['0-3k', '3k-6k', '6k-9k', '9k-12k', '12k-15k', '15k+'])

df['AgeGroup'] = pd.cut(df['Age'],
                        bins=[18, 25, 35, 45, 60],
                        labels=['18-25','26-35','36-45','46-60'])

df['YearsAtCompanyGroup'] = pd.cut(df['YearsAtCompany'],
                                    bins=[0, 2, 5, 10, 20, 40],
                                    labels=['0-2','3-5','6-10','11-20','21+'])
```

```python
# Attrition Rate
attrition_counts = df['Attrition'].value_counts()
attrition_rate = (attrition_counts['Yes']/attrition_counts.sum())*100
print(f"Overall Attrition Rate : {attrition_rate:.2f}%")

plt.figure(figsize=(6,4))
sns.countplot(x='Attrition', data=df, palette=['grey', 'skyblue'])
plt.title(f"Overall Attrition Rate: {attrition_rate:.1f}%")
plt.show()
```

```python
# Attrition By Age
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='Age', hue='Attrition', kde=True, bins=20, multiple='stack' )
plt.title("Attrition by Age")
plt.show()
```

```python
# Attrition by Years at Company
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='YearsAtCompany', hue='Attrition', kde=True, bins=20, multiple='stack')
plt.title("Attrition by Years at Company")
plt.show()
```

```python
# Attrition By Monthly Income
plt.figure(figsize=(10,6))
sns.countplot(x='IncomeBin', hue='Attrition', data=df, palette='Blues')
plt.title("Attritioin By Monthly Income")
plt.xticks(rotation=45)
plt.show()
```

```python
# Attrition By Age Group
plt.figure(figsize=(10,6))
sns.countplot(x='AgeGroup', hue='Attrition', data=df, palette='mako')
plt.title("Attrition by Age Group")
plt.show()
```

```python
# Attrition by Gender
LeftEmployees = df[df['Attrition']=='Yes']
gender_pie = LeftEmployees['Gender'].value_counts()

# pie chart
plt.figure(figsize=(6,6))
plt.pie(gender_pie, labels=gender_pie.index, autopct='%1.1f%%', colors=['lightcoral','lightskyblue'])
plt.title("Gender Distribution Among Employees who Left")
plt.axis('equal')
plt.show()
```

```python
# Attrition By Education Level 
edu_map = {
    1: 'Below College', 2:'College', 3:'Bachelor', 4:'Master',5:'Doctor'
}
if df['Education'].dtype in ['int64','float64']:
    df['EducationLevel'] = df['Education'].map(edu_map)

plt.figure()
sns.countplot(x='EducationLevel',hue='Attrition',data=df,palette='Set2')
plt.title("Attrition By Education Level")
plt.xticks(rotation=45)
plt.show()
```

```python
# Years since Last Promotion
plt.figure(figsize=(10,6))
sns.histplot(data=df, x='YearsSinceLastPromotion', hue='Attrition', multiple='stack', bins=15, palette='Set2')
plt.title("Distribution of Years Since Last Promotion by Attrition Status")
plt.xlabel("Years Since Last Promotion")
plt.ylabel("Number of Employees")
plt.show()
```

```python
# Attrition by Overtime 
plt.figure()
sns.countplot(x='OverTime', hue='Attrition', data=df, palette='YlGnBu')
plt.title("Attrition by OverTime Status")
plt.show()
```

```python
# Attrition by Job Role And Environment Satisfaction
plt.figure(figsize=(14,6))
sns.barplot(data=df, x='EnvironmentSatisfaction', y='JobRole', hue='Attrition', errorbar=None, palette='Set2')
plt.title("Average Environment Satisfaction per Job Role (Grouped by Attrition)")
plt.xlabel("Average Environment Satisfaction")
plt.ylabel("Job Role")
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.legend(title='Attrition')
plt.show()
```

```python
# Correlation with Attrition
corr_matrix = df[['Age', 'YearsAtCompany', 'MonthlyIncome', 'Attrition_Encoded']].corr()

# Plot heatmap
plt.figure(figsize=(8,6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title("Correlation Matrix (Including Attrition)")
plt.show()
```
